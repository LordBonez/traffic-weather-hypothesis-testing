```python
import pandas as pd
import numpy as np
merged_df = pd.read_csv('../data/merged_traffic_weather.csv')
merged_df['total_vol_log'] = np.log1p(merged_df['total_vol'])
```

# 2. Hypothesis Testing

*   Null Hypothesis (H0) - Weather conditions do not affect the traffic volume in cities.

*   Alternative Hypothesis (HA) - Weather conditions do affect the traffic volume in cities.









```python
#T-Test

from scipy import stats
import numpy as np

merged_df['rain_flag'] = np.where(merged_df['precipitation (mm)'] > 0, "Rain", "Dry")

rain = merged_df[merged_df['rain_flag'] == "Rain"]['total_vol_log']
dry = merged_df[merged_df['rain_flag'] == "Dry"]['total_vol_log']

t_stat, p_value = stats.ttest_ind(rain, dry, equal_var=False)

print("T-statistic:", t_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Reject the null hypothesis. There is evidence of a difference in mean log traffic volume between rainy and dry hours.")
else:
    print("Fail to reject the null hypothesis. There is insufficient evidence to reject equal mean log traffic volume between the groups.")
```

At alpha = 0.05, a small p-value provides evidence against equal mean log traffic volume between rainy and dry hours.
A non-significant result is not proof that the means are equal. Statistical significance alone does not establish practical importance or a causal weather effect. Hourly dependence and time-of-day confounding limit this simple group comparison.


```python
#ANOVA

def categorize_weather(row):
    if row['precipitation (mm)'] > 0:
        return "Rain"
    else:
        return "Dry"

merged_df['weather_category'] = merged_df.apply(categorize_weather, axis=1)
merged_df['weather_category'].value_counts()


groups = [group['total_vol_log'].values
          for name, group in merged_df.groupby('weather_category')]

f_stat, p_value = stats.f_oneway(*groups)

print("F-statistic:", f_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Reject the null hypothesis. There is evidence of a difference in mean log traffic volume between rainy and dry hours.")
else:
    print("Fail to reject the null hypothesis. There is insufficient evidence to reject equal mean log traffic volume between the groups.")
```
